## 2.Teoria 

Përshtatja lineare (Linear Regression) përdoret për të gjetur marrëdhënien midis dy madhësive. Supozohet se të dhënat mund të përshkruhen nga një ekuacion i formës:

$$
y=ax+b
$$

ku:

* ($a$) është pjerrësia e vijës,
* ($b$) është prerja me boshtin ($y$).

###  5.2. Gjenerimi i të dhënave

Të dhënat janë gjeneruar duke përdorur skriptin:


```python
import numpy as np
import os

# ID e studentit
username = "Albana-1"

seed = sum(ord(c) for c in username)
np.random.seed(seed)

# Parametrat linearë
a = 2 + (seed % 3)     # pjerrësia
b = 5 + (seed % 5)     # prerja

# Të dhënat
x = np.linspace(1, 10, 20)
noise = np.random.normal(0, 1, size=len(x))
y = a * x + b + noise

# Ruajtja
os.makedirs("data", exist_ok=True)
np.savetxt("data/model.dat", np.column_stack((x, y)),
           delimiter=",", header="x,y", comments="")

print("model.dat u krijua!")
```

Ky skript krijon një dataset unik për çdo student duke përdorur:
- Parametra të ndryshëm linearë (a dhe b)
- Zhurmë të rastësishme (noise)

Dataset-i është:
- Unik për çdo student
- I riprodhueshëm (me të njëjtën ID)

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt("data/model.dat", delimiter=",", skiprows=1)

x_Data = data[:, 0]
y_Data = data[:, 1]
```

###  5.3. Përshtatja lineare

Për të gjetur vijën që përshkruan më mirë të dhënat, përdorim regresion linear.

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt("data/model.dat", delimiter=",", skiprows=1)

x_Data = data[:, 0]
y_Data = data[:, 1]
a, b = np.polyfit(x_Data, y_Data, 1)

print(f"Pjerrësia a = {a:.2f}")
print(f"Prerja b = {b:.2f}")
```

### 5.4. Vizualizimi i të dhënave dhe modelit

Grafiku i mëposhtëm paraqet të dhënat eksperimentale dhe vijën e përshtatjes lineare.

```python
xfit = np.linspace(1, 10, 100)
yfit = a * xfit + b

fig=plt.figure(figsize=(8,6))
plt.plot(x_Data, y_Data, 'ok', label='Të dhënat')
plt.plot(xfit, yfit, '-r', lw=2, label=f'y = {a:.2f}x + {b:.2f}')
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()

fig.savefig("figs/histo.png")

plt.show()
```

## 6. Analiza e mbetjeve (Residuals)

Mbetjet janë diferenca midis vlerave reale dhe atyre të modelit.
Ato ndihmojnë në vlerësimin e cilësisë së regresionit.

```python
residuals = y_Data - (a * x_Data + b)

fig=plt.figure(figsize=(8,6))
plt.stem(x_Data, residuals)
plt.axhline(0, linestyle='--')
plt.xlabel("x")
plt.ylabel("Mbetjet")
plt.xlim(0.8, 10.2)
plt.grid()

fig.savefig("figs/histo1.png")

plt.show()
```

##  7. Interpretimi i rezultateve

- Pjerrësia (a) tregon lidhjen lineare midis x dhe y.
- Prerja (b) tregon vlerën e y kur x = 0.

Në grafikun e parë shohim që modeli linear përshtatet mirë me të dhënat.

Në grafikun e mbetjeve:
- Mbetjet janë të shpërndara rreth zeros
- Nuk vërehet ndonjë model i qartë

Kjo tregon se modeli linear është i përshtatshëm për këtë dataset.

## 8.Pasiguria

Të dhënat përmbajnë zhurmë të rastësishme (noise), e cila simulon pasiguritë eksperimentale që hasen në matje reale.

Burimet kryesore të pasigurisë janë:

* Zhurma e shtuar gjatë gjenerimit të të dhënave.
* Gabimet numerike gjatë llogaritjeve.
* Devijimet e pikave nga modeli ideal linear.

Pasiguria vlerësohet duke analizuar mbetjet (residuals). Sa më të vogla dhe më të shpërndara rastësisht rreth zeros të jenë mbetjet, aq më i mirë është modeli.

 ## 9.Diskutimi 

 Rezultatet tregojnë se ekziston një lidhje afërsisht lineare midis madhësive (x) dhe (y). Vlerat e llogaritura për pjerrësinë dhe prerjen përshkruajnë mirë të dhënat eksperimentale.

Nga grafiku i mbetjeve vërehet se shumica e pikave janë të shpërndara rreth zeros pa ndonjë model të veçantë. Kjo tregon se devijimet vijnë kryesisht nga zhurma rastësore dhe jo nga një gabim sistematik i modelit.

Megjithatë, për shkak të pranisë së zhurmës, modeli nuk përputhet në mënyrë perfekte me të gjitha pikat. Kjo përfaqëson pasiguritë që do të haseshin edhe në një eksperiment real.
