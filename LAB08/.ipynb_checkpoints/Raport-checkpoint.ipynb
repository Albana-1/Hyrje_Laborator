## 1.Qëllimi i Laboratorit
Ky laborator synon kuptimin e shumë madhësive fizike që nuk maten drejtpërdrejt, por llogariten nga madhësi të tjera të matura. 

Objektivat janë:
* Të matet masa dhe përmasat gjeometrike
* Të llogaritet vëllimi
* Të përcaktohet dendësia
* Të kuptohet dhe zbatohet përhapja e pasigurisë
* Të vlerësohet se cila matje dominon pasigurinë finale


## 2.Hyrje Teorike

### 2.1. Matja direkte dhe indirekte

* Matje direkte: madhësia lexohet direkt nga instrumenti p.sh. masa me peshore, gjatësia me kaliper

* Matje indirekte: madhësia llogaritet nga madhësi të tjera të matura p.sh. dendësia: $$ \rho = \frac{m}{V} $$

Pra, dendësia nuk lexohet direkt; ajo del nga matja e masës dhe vëllimit.


### 2.2.Dendësia
Dendësia përkufizohet si masa për njësi vëllimi:

$$ \rho = \frac{m}{V} $$

Njësia SI është:

$$ [\rho] = \mathrm{kg/m^3} $$

Në laborator shpesh përdoret edhe:

$$ \mathrm{g/cm^3} $$

### 2.3.Pse nevojitet përhapja e pasigurisë?
Nëse (m) dhe (V) kanë pasiguri, atëherë edhe (\rho) do të ketë pasiguri. Pra nuk mjafton të japim vetëm vlerën e dendësisë; duhet të japim edhe sa e sigurt është ajo vlerë.

### Konceptet Kryesore
####  **Pasiguria absolute**
Nëse një madhësi matet si:
$$ x = x_0 \pm \Delta x $$
atëherë $\Delta x$ është pasiguria absolute.
#### **Pasiguria relative**
$$ \frac{\Delta x}{x} $$
Shpesh shprehet në përqindje:
$$ \frac{\Delta x}{x}\times 100% $$

## 7.Analiza e të Dhënave

* Gjejmë mesataren me formulën:

   $$ \bar{m} = \frac{1}{N}\sum_i m_i $$


   $$ \bar{m} = \frac{175}{4} = 43.75  $$
   
* Gjejmë vëllimin  me formulën:

$$ V=abc $$


$ V_1 = 39.5 \cdot 3 \cdot 3.65 = 432.5 \, \text{cm}^3 $

$ V_2 = 39.5 \cdot 3 \cdot 6.45 = 764.3 \, \text{cm}^3 $

$ V_3 = 39.5 \cdot 3 \cdot 10.13 = 1200.4 \, \text{cm}^3 $

$ V_4 = 39.5 \cdot 3 \cdot 12.74 = 1509.7 \, \text{cm}^3 $

* Gjejmë dendësinë me formulën:
  
$$ \rho = \frac{m}{V} $$

$ \rho_1 = \frac{43}{432.5} = 0.099 \, \frac{\text{g}}{\text{cm}^3}  $

$ \rho_2 = \frac{87}{764.3} = 0.114 \, \frac{\text{g}}{\text{cm}^3} $

$ \rho_3 = \frac{134}{1200.4} =0.112 \, \frac{\text{g}}{\text{cm}^3}  $

$ \rho_4 = \frac{12.74}{1509.7} = 0.008 \,\frac{\text{g}}{\text{cm}^3}  $


* Gjejmë pasiguritë relative të madhësive bazë:

$$ \frac{\Delta V}{V} \approx \frac{\Delta a}{a} + \frac{\Delta b}{b} + \frac{\Delta c}{c}$$

Për $V_1$:

$ \frac{\Delta V_1}{V_1} =
\frac{0.1}{39.5} + \frac{0.1}{3} + \frac{0.01}{3.65} = 0.0385 \approx 3.85\% $

$ \frac{\Delta V_2}{V_2} \approx 3.75\% $

$ \frac{\Delta V_3}{V_3} \approx 3.70\% $

$ \frac{\Delta V_4}{V_4} \approx 3.69\% $


$$ \frac{\Delta \rho}{\rho} \approx \frac{\Delta m}{m} + \frac{\Delta V}{V} $$



Për $ \rho_1 $:

$ \frac{\Delta \rho_1}{\rho_1} =
\frac{1}{43} + 0.0385 = 0.0615 \approx 6.15\% $

Për $ \rho_2 $:

$ \frac{\Delta \rho_2}{\rho_2} =
\frac{1}{87} + 0.0375 = 0.049 \approx 4.9\% $

Për $ \rho_3 $:

$ \frac{\Delta \rho_3}{\rho_3} =
\frac{1}{134} + 0.037 = 0.0445 \approx 4.45\% $

Për $ \rho_4 $:

$ \frac{\Delta \rho_4}{\rho_4} =
\frac{1}{175} + 0.0369 = 0.0426 \approx 4.26\% $

* Kthejmë në pasiguri absolute:

 $$ \Delta V = V \left( \frac{\Delta a}{a} + \frac{\Delta b}{b} + \frac{\Delta c}{c} \right) $$

* $ V_1 = (432.5 \pm 16.7) \, \text{cm}^3 $

* $ V_2 = (764.3 \pm 28.7) \, \text{cm}^3 $

* $ V_3 = (1200.4 \pm 44.4) \, \text{cm}^3 $

* $ V_4 = (1509.7 \pm 55.7) \, \text{cm}^3 $

$$  \rho = (\rho_0 \pm \Delta \rho) $$ 

* $ \rho_1 = (0.099 \pm 0.006) \, \frac{\text{g}}{\text{cm}^3} $

* $ \rho_2 = (0.114 \pm 0.006) \, \frac{\text{g}}{\text{cm}^3} $

* $ \rho_3 = (0.112 \pm 0.005) \, \frac{\text{g}}{\text{cm}^3} $

* $ \rho_4 = (0.116 \pm 0.005) \, \frac{\text{g}}{\text{cm}^3} $

```python
import numpy as np

m = np.array([43, 87, 134, 175])
print("Masa mesatare =", np.mean(m), "g")
```

```python
a = np.array([39.5, 39.5, 39.5, 39.5])
b = np.array([3, 3, 3, 3])
c = np.array([3.65, 6.45, 10.13, 12.74])

V = a * b * c
print("Vëllimet:", V)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Te dhenat
V = [432.5, 764.3, 1200.4, 1509.7]  # cm^3
m = [43, 87, 134, 175]              # g


fig=plt.figure(0,(5,4))  
plt.plot(V, m, '.-k', ms=10, ) 
plt.plot(V, m, '-r', lw=1,) 
plt.xlabel("Vëllimi (cm³)")
plt.ylabel("Masa (g)")
plt.title(r'$m$ vs. $V$ ')   
plt.grid()
plt.scatter(V, m)

fig.savefig("figs/histo.png")

plt.show()
```
