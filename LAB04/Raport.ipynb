## **6. Analiza**

Pas përpunimit të të dhënave:

```python
import numpy as np
import matplotlib.pyplot as plt

te_dhena = np.array([10.1, 10.2, 9.9, 10.0, 10.3, 10.1, 10.2, 10.0, 9.8, 
                     10.1, 10.2, 10.1, 10.0, 9.9, 10.2])

# Llogaritja e  mesatares aritmetike
mesatarja = np.mean(te_dhena)

print(f"Mesatarja: {mesatarja:.2f}")
```

Grafikët vizualë përfshinë:

* Histogramin, për të parë shpërndarjen e matjeve.
* Grafikun sipas rendit, për të analizuar ndryshimet gjatë eksperimentit.

Grafikët tregojnë që shumica e matjeve janë të përqendruara afër mesatares dhe nuk ka vlera ekstreme.

```python
import numpy as np
import matplotlib.pyplot as plt

te_dhena = np.array([10.1, 10.2, 9.9, 10.0, 10.3, 10.1, 10.2, 10.0, 9.8, 
                     10.1, 10.2, 10.1, 10.0, 9.9, 10.2])

# Llogaritja e mesatares aritmetike
mesatarja = np.mean(te_dhena)

print(f"Mesatarja: {mesatarja:.2f}")

# Ndërtimi i Histogramës
plt.hist(te_dhena, bins=5, color='skyblue', edgecolor='black')
plt.title("Histogram i të dhënave")
plt.xlabel("Vlera")
plt.ylabel("Numri i herëve")
plt.show()
```
