## 4. Analiza Statistikore
4.1 Nga Distancat
Mesatarja e distancës:
$\bar{d} \approx 15.21 \,\text{cm}$
Koha mesatare:
$\bar{t} \approx 0.176 \,\text{s}$
Devijimi standard:
$\sigma \approx 0.03 \, s$
4.2 Nga Koha e Matur

Mesatarja:

$$
\bar{t} = \frac{1}{N} \sum t_i
$$

$$
\bar{t} \approx 0.146 \, s
$$

Devijimi standard:

$$
\sigma = \sqrt{\frac{1}{N-1} \sum (t_i - \bar{t})^2}
$$

$$
\sigma \approx 0.074 \, s
$$

5. Krahasimi i Rezultateve
| Metoda           | Mesatarja (s) | Devijimi Standard (s) |
|------------------|--------------|------------------------|
| Nga distanca     | 0.176        | 0.03                   |
| Matje direkte    | 0.146        | 0.074                  |
7. Interpretimi
* Koha e reagimit nuk është konstante
* Ekziston një shpërndarje e qartë e vlerave
* Matjet direkte kanë variacion më të madh
* Ka vlera ekstreme që ndikojnë rezultatin
* Shpërndarja është afërsisht normale
8. Rezultatet Finale

Nga matjet direkte:

$$
t = (0.146 \pm 0.074)\, s
$$

Nga metoda me distancë:

$$
t = (0.176 \pm 0.03)\, s
$$


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# 1. Matjet e nxjerra  (në cm)
distancat_cm = [
   27.3,17.5,24.6,12.8,13,18.5,25,12.5,18.3,18,16.5,15,
13.5,9.6,21,9,10,12.5,16.9,15.3,12,11.9,11.5,18.7,
12.5,15.5,12,12,19.4,12.9,14.5,14,16.5,13.5,11.6,12.3
]

# 2. Konvertimi në metra dhe llogaritja e kohës t = sqrt(2d/g)
d = np.array(distancat_cm) / 100
g = 9.81
t = np.sqrt((2 * d) / g)

# 3. Llogaritjet statistikore
N = len(t)
mesatarja = np.mean(t)
std_dev = np.std(t, ddof=1)

print(f"Numri i matjeve: {N}")
print(f"Mesatarja (t_bar): {mesatarja:.4f} s")
print(f"Devijimi Standard (sigma): {std_dev:.4f} s")
print(f"Rezultati Final: t = ({mesatarja:.3f} ± {std_dev:.3f}) s")

# 4. Ndërtimi i grafikut (Histogrami + Kurba Gaussiane)
plt.figure(figsize=(10, 6))

# Histogrami i normalizuar 
count, bins, ignored = plt.hist(t, bins=10, density=True, alpha=0.6, color='skyblue', edgecolor='black', label='Të dhënat eksperimentale')

# Kurba teorike Gaussiane
x = np.linspace(min(t) - 0.05, max(t) + 0.05, 100)
p = norm.pdf(x, mesatarja, std_dev)
plt.plot(x, p, 'r', linewidth=2, label='Shpërndarja Normale (Gaussian)')

plt.title(' Shpërndarja e Kohës së Reagimit (Hedhja e vizores)')
plt.xlabel('Koha (s)')
plt.ylabel('Densiteti i Probabilitetit')
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.show()
```
