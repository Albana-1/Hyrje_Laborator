## **Laboratori 03: Menaxhimi i Versioneve me Git dhe GitHub**

### **Qëllimi i Laboratorit**

Qëllimi i këtij laboratori është të paraqesë në mënyrë të strukturuar dhe praktike përdorimin e sistemit të kontrollit të versioneve Git, si dhe integrimin e tij me platformën GitHub për publikimin dhe bashkëpunimin në projekte zhvillimi.

---

### **Hyrje Teorike**

**Git** është një sistem i shpërndarë për kontrollin e versioneve që mundëson ruajtjen dhe gjurmimin e ndryshimeve në skedarë përgjatë zhvillimit të një projekti.

**GitHub** është një platformë në cloud që ofron hostim për repositorët Git dhe lehtëson bashkëpunimin ndërmjet zhvilluesve përmes funksionaliteteve të avancuara.

---

### **Konfigurimi Fillestar**

Në fillim të përdorimit të Git, është e nevojshme konfigurimi i të dhënave të përdoruesit në nivel lokal:


`!git config --global user.name "Emri Juaj"`

`!git config --global user.email "email@shembull.com"`


Verifikimi i konfigurimit kryhet përmes komandës:


`!git config --list | grep user`

---

### **Inicializimi i Repositorit**

Për të filluar menaxhimin e një projekti me Git, inicializohet një repositor lokal:


`!git init`

Gjendja aktuale e repositorit mund të kontrollohet me:


`!git status`


---

### **Menaxhimi i Ndryshimeve (Add dhe Commit)**

Krijohet një skedar i ri për testim:


!echo "Ky është një shembull testimi" > test.py


Shtimi i skedarit në staging area realizohet me:


`!git add test.py`

Ruajtja e ndryshimeve në repositorin lokal bëhet përmes commit-it:


`!git commit -m "LAB 03: Shtimi i skedarit test.py"`

Historiku i veprimeve mund të shfaqet me:


`!git log --oneline`

---

### **Publikimi në GitHub**

Për të publikuar projektin në GitHub, repositori lokal lidhet me një repositor të largët (remote):

`!git remote add origin https://github.com/username/repository.git`

`!git push -u origin main`

---

### **Përfundime dhe Reflektime**

Gjatë zhvillimit të këtij laboratori u arritën këto përfundime kryesore:

1. **Rëndësia e Kontrollit të Versioneve**
   Git mundëson ruajtjen e çdo ndryshimi në projekt dhe ofron siguri për rikthimin në versione të mëparshme në rast gabimesh.

2. **Dallimi ndërmjet Git dhe GitHub**
   Git funksionon si mjet lokal për menaxhimin e kodit, ndërsa GitHub shërben si platformë për publikimin dhe bashkëpunimin në projekte.

3. **Zbatimi i Workflow-it Standard**
   Procesi i ndjekur (edit → add → commit → push) përbën praktikën standarde në zhvillimin e softuerit dhe është thelbësor për organizimin dhe bashkëpunimin efektiv.

---

### **Statusi i Detyrës**

Laboratori u përfundua me sukses dhe projekti u publikua në platformën GitHub.
