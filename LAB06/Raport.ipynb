## $$ x = x_{\text{mat}} \pm \Delta x $$1. Qëllimi
Qëllimi i këtij eksperimenti është të maten dimensionet e disa objekteve (kuti, cilindër, rreth dhe telefon) duke përdorur instrumente të ndryshme matëse dhe të krahasohet saktësia e tyre përmes analizës së gabimeve.

Gjatësia është madhësi fizike bazë që mat distancën midis dy pikave.

Njësia SI është metri (m), por në laborator përdoren cm dhe mm.

Gabimi:
$$ x = x_{\text{mat}} \pm \Delta x $$

## 2.Teoria

Gabimi në matje mund të vijë nga:

* kufizimi i instrumentit
* leximi i gabuar i shkallës

## Llogaritja e pasigurisë

| Objekti | Madhësia | Mesatarja (x̄) | Pasiguria (Δx) |
|----------|-----------|----------------|----------------|
| Kuti     | a         | 131.98 mm      | ±0.95 mm       |
| Kuti     | b         | 57.69 mm       | ±1.28 mm       |
| Peshë    | d         | 29.33 mm       | ±1.00 mm       |
| Peshë    | l         | 40.06 mm       | ±0.09 mm       |
| Rreth    | l         | 40.13 mm       | ±0.10 mm       |
| Rreth    | t         | 1.726 mm       | ±0.014 mm      |
| Telefon  | a         | 160.40 mm      | ±0.40 mm       |
| Telefon  | b         | 76.65 mm       | ±1.65 mm       |


$$
\bar{x} = \frac{x_1 + x_2 + x_3 + \cdots + x_n}{n}
$$

### Ku:

- $\bar{x}$ = mesatarja  
- $x_i$ = vlerat e matjeve  
- $n$ = numri i matjeve

$$
x = \bar{x} \pm \Delta x
$$

### Ku:

- $x_{\max}$ = vlera më e madhe  
- $x_{\min}$ = vlera më e vogël  
- $\Delta x$ = pasiguria absolute  

```python
import matplotlib.pyplot as plt

labels = ["Kuti a", "Kuti b", "Peshë d", "Peshë l", "Telefon a", "Telefon b"]
values = [130, 56, 28, 40, 160, 75]

fig=plt.figure()

plt.plot(labels, values, marker='o', color="#1f77b4", label="Vizore")

plt.title("Vizore - Matjet")
plt.ylabel("mm")
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

plt.legend()

plt.tight_layout()

fig.savefig("figs/Matje me vizore.png")
plt.show()
```
